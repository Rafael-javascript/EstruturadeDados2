```python
rede = {
    "Rafael": ["Júlio", "Alisson", "Laís", "Pedro", "Mariana"],
    "Júlio": ["Rafael", "Alisson", "Diego", "Mariana"],
    "Alisson": ["Rafael", "Júlio", "Diego"],
    "Pedro": ["Mariana", "Rafael"],
    "Diego": ["Júlio", "Alisson"],
    "Mariana": ["Pedro", "Júlio", "Rafael"],
    "Laís": ["Rafael"],
}

# Quantidade de usuarios (passamos o dicionario 'rede')
n = len(rede)

# Calcule o grau de cada usuario (usamos 'len(amigos)' para cada pessoa)
for usuario, amigos in rede.items():
    print(usuario, "tem", len(amigos), "conexoes")

# A soma dos graus conta cada amizade duas vezes
soma_graus = sum(len(amigos) for amigos in rede.values())
arestas = soma_graus // 2

# Imprime o total usando as variaveis calculadas (n e arestas)
print("Total de usuarios:", n)
print("Total de amizades:", arestas)

# Densidade do grafo
densidade = (2 * arestas) / (n * (n - 1))
print(f"Densidade: {densidade:.2%}")
```
